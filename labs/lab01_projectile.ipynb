```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
g = 9.81         # m/s², gravitational acceleration
dt = 0.01        # time step in seconds
t_max = 4.0      # total simulation time

# Initial conditions: launched at 45°, speed 20 m/s
v0 = 20.0
angle = np.radians(45)
x, y = 0.0, 0.0
vx, vy = v0 * np.cos(angle), v0 * np.sin(angle)

# Storage lists for plotting
xs, ys, ts = [x], [y], [0.0]

# Euler integration loop
t = 0.0
while t < t_max and y >= 0.0:
    # Compute acceleration (only gravity here)
    ax = 0.0
    ay = -g

    # Update velocity (Euler)
    vx = vx + ax * dt
    vy = vy + ay * dt

    # Update position (Euler)
    x = x + vx * dt
    y = y + vy * dt

    t += dt
    xs.append(x)
    ys.append(y)
    ts.append(t)

# Plot trajectory
plt.figure(figsize=(8, 4))
plt.plot(xs, ys, 'b-', linewidth=2)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Projectile Trajectory — Euler Method')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# Physical parameters

g = 9.81
m = 0.145          # baseball mass (kg)
r = 0.0366         # baseball radius (m)
Cd = 0.47          # drag coefficient
rho = 1.225        # air density (kg/m^3)
A = np.pi * r**2

# Drag constant
b = 0.5 * Cd * rho * A

# Simulation parameters
dt = 0.01
v0 = 40.0
angle = np.radians(45)








# TRAJECTORY WITHOUT AIR RESISTANCE


x, y = 0.0, 1.0

vx = v0 * np.cos(angle)
vy = v0 * np.sin(angle)

xs_no_drag = [x]
ys_no_drag = [y]

while y >= 0.0:

    # Acceleration from gravity
    ax = 0.0
    ay = -g

    # Update velocity
    vx += ax * dt
    vy += ay * dt

    # Update position
    x += vx * dt
    y += vy * dt

    xs_no_drag.append(x)
    ys_no_drag.append(y)











# TRAJECTORY WITH AIR RESISTANCE


# Reset initial conditions
x, y = 0.0, 1.0

vx = v0 * np.cos(angle)
vy = v0 * np.sin(angle)

xs_drag = [x]
ys_drag = [y]

while y >= 0.0:

    # Magnitude of velocity
    speed = np.sqrt(vx**2 + vy**2)

    # Acceleration from gravity + drag
    ax = -(b / m) * speed * vx
    ay = -g - (b / m) * speed * vy

    # Update velocity
    vx += ax * dt
    vy += ay * dt

    # Update position
    x += vx * dt
    y += vy * dt

    xs_drag.append(x)
    ys_drag.append(y)



# PLOT BOTH TRAJECTORIES

plt.figure(figsize=(10, 5))

plt.plot(xs_no_drag, ys_no_drag,
         label='Without air resistance')

plt.plot(xs_drag, ys_drag,
         label='With air resistance')

plt.xlabel('Horizontal Distance (m)')
plt.ylabel('Height (m)')
plt.title('Projectile Motion With and Without Air Resistance')

plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(bottom=0)

plt.show()
```
